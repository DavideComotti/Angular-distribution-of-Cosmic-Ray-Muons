```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1 = pd.read_csv('/content/drive/MyDrive/Terzo anno/Tirocinio/Python/Mixed data- Legnaro/File uniti/File/SL_filtrato.csv')

df2 = pd.read_csv('/content/drive/MyDrive/Terzo anno/Tirocinio/Python/Mixed data- Legnaro/File uniti/File/TF_filtrato.csv')

print(df1)
print(df2)
```

```python
#Super Layer
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Calcolo angolo in gradi e valore assoluto
msl0 = np.arctan(df1.m_SL_0)
msl0 = np.degrees(msl0)
msl0 = np.abs(msl0)
msl0 = msl0[~np.isnan(msl0)]

# Istogramma con frequenza relativa percentuale
counts, bin_edges = np.histogram(msl0, bins=150)
counts_percent = (counts / counts.sum()) * 100

# Disegna l'istogramma a barre
plt.figure(figsize=(6, 5))
plt.bar(bin_edges[:-1], counts_percent, width=np.diff(bin_edges), align='edge')

plt.grid(True)
plt.xlabel('Angle [°]')
plt.xlim(0, 90)
plt.ylabel('Relative frequency [%]')
plt.title('Super layer: Vertical chamber (Frequency % Relative)')
plt.tight_layout()
plt.show()
```
