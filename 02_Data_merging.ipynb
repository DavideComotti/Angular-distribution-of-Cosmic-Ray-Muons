```python
#1
import uproot
import pandas as pd

array1 = uproot.concatenate(
    {path: "TFTree" for path in file_paths1},
    filter_name=["timestamp", "m_SL", "m_TF", "hasMT"],
    library="pd")
```

```python
#2
import uproot
import pandas as pd

array2 = uproot.concatenate(
    {path: "TFTree" for path in file_paths2},
    filter_name=["timestamp", "m_SL", "m_TF", "hasMT"],
    library="pd")
```

```python
#3
import uproot
import pandas as pd

array3 = uproot.concatenate(
    {path: "TFTree" for path in file_paths3},
    filter_name=["timestamp", "m_SL", "m_TF", "hasMT"],
    library="pd")
```

```python
#4
import uproot
import pandas as pd

array4 = uproot.concatenate(
    {path: "TFTree" for path in file_paths4},
    filter_name=["timestamp", "m_SL", "m_TF", "hasMT"],
    library="pd")
```

```python
final_array = pd.concat([array1, array2, array3, array4], ignore_index=True)
```

```python
# Espandi 'm_TF'
final_array[["m_TF_0", "m_TF_1"]] = pd.DataFrame(final_array["m_TF"].tolist(), index=final_array.index)

# Espandi 'm_SL'
final_array[["m_SL_0", "m_SL_1"]] = pd.DataFrame(final_array["m_SL"].tolist(), index=final_array.index)

# Espandi 'hasMT'
final_array[["hasMT_0", "hasMT_1"]] = pd.DataFrame(final_array["hasMT"].tolist(), index=final_array.index)

# Rimuovi le colonne originali
final_array.drop(columns=["m_TF", "m_SL", "hasMT"], inplace=True)
```
