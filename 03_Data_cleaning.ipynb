```python
#File contenente m_SL, hasMT, timestamp

import pandas as pd

# Carica colonne necessarie
df1 = pd.read_csv("/content/drive/MyDrive/Terzo anno/Tirocinio/Python/Mixed data- Legnaro/File uniti/File/mixed_data_ali_TOTALE.csv", usecols=["timestamp", "m_SL_0", "m_SL_1", "hasMT_0", "hasMT_1"])

# Maschere per i due canali separati
mask0 = (df1["m_SL_0"] > -9999) & (df1["hasMT_0"] == True)
mask1 = (df1["m_SL_1"] > -9999) & (df1["hasMT_1"] == True)

# Estrai i valori validi per m_SL_0 e timestamp0
m_SL_0_valid = df1.loc[mask0, "m_SL_0"].reset_index(drop=True)
timestamp0 = df1.loc[mask0, "timestamp"].reset_index(drop=True)

# Estrai i valori validi per m_SL_1 e timestamp1
m_SL_1_valid = df1.loc[mask1, "m_SL_1"].reset_index(drop=True)
timestamp1 = df1.loc[mask1, "timestamp"].reset_index(drop=True)

# Crea il DataFrame finale con timestamp duplicati, ciascuno filtrato sul suo canale
df2 = pd.DataFrame({
    "m_SL_0": m_SL_0_valid,
    "timestamp0": timestamp0,
    "m_SL_1": m_SL_1_valid,
    "timestamp1": timestamp1
})

print(df2.head())
print(len(df2.m_SL_0))
```

```python
df2.to_csv("SL_filtrato.csv", index=False)
```
